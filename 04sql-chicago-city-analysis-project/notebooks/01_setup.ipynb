```python
import pandas as pd
import sqlite3

con = sqlite3.connect("chicago_analysis.db")

#Loading tables

datasets = {
    "CENSUS_DATA": "https://data.cityofchicago.org/resource/kn9c-c2s2.csv",
    "CHICAGO_PUBLIC_SCHOOLS_DATA": "https://data.cityofchicago.org/resource/9xs2-f89t.csv",
    "CRIME_DATA":"https://data.cityofchicago.org/resource/crimes.csv?$query=SELECT%0A%20%20%60id%60%2C%0A%20%20%60primary_type%60%2C%0A%20%20%60description%60%2C%0A%20%20%60location_description%60%2C%0A%20%20%60arrest%60%2C%0A%20%20%60domestic%60%2C%0A%20%20%60community_area%60%2C%0A%20%20%60year%60%0AWHERE%20%60year%60%20BETWEEN%202011%20AND%202012%0AORDER%20BY%20%60date%60%20DESC%20NULL%20FIRST%20LIMIT%20700000"
}

for table, url in datasets.items():
    df=pd.read_csv(url)
    df.to_sql(table, con, if_exists='replace', index=False, chunksize=1000)
    print(f" {table} loaded — {len(df)} rows")

con.commit()
```

```python
%%sql

CREATE TABLE CHICAGO_PUBLIC_SCHOOLS_CLEANED AS
SELECT
    school_id,
    name_of_school,
    elementary_or_high_school,
    community_area_number,
    community_area_name,

    safety_score,
    family_involvement_score,

    average_student_attendance,
    average_teacher_attendance,
    rate_of_misconducts_per_100_students_,

    graduation_rate_,
    college_eligibility_,
    college_enrollment_rate,
    college_enrollment_number_of_students_,
    freshman_on_track_rate,

    isat_exceeding_math_,
    isat_exceeding_reading_

FROM CHICAGO_PUBLIC_SCHOOLS_DATA
```
